```python
from PIL import Image

ASCII_CHARS = "@%#*+=-:. "
WIDTH = 80

def process_image(image_path, new_width=WIDTH):
    try:
        image = Image.open(image_path)
        width, height = image.size
        aspect_ratio = height / width / 1.65
        new_height = int(new_width * aspect_ratio)
        image = image.resize((new_width, new_height))
        image = image.convert('L')
        pixels = image.getdata()
        ascii_str = ''.join(ASCII_CHARS[pixel // 25] for pixel in pixels)
        ascii_img = '\n'.join(ascii_str[i:i+new_width] for i in range(0, len(ascii_str), new_width))
        print(ascii_img)
        with open('ascii_image.txt', 'w') as f:
            f.write(ascii_img)
    except Exception as e:
        print(f"Unable to process image: {e}")

image_path = 'fly.jpg'
process_image(image_path)
```

```python
from PIL import Image

ASCII_CHARS = "@%#*+=-:. "  # Define the characters used for ASCII representation
WIDTH = 80  # Define the width for the ASCII art

def process_image(image_path, new_width=WIDTH):
    try:
        image = Image.open(image_path)  # Open the image file
        width, height = image.size
        aspect_ratio = height / width  # Correct the aspect ratio calculation
        new_height = int(new_width * aspect_ratio)  # Calculate new height based on aspect ratio
        image = image.resize((new_width, new_height))  # Resize the image
        image = image.convert('L')  # Convert the image to grayscale
        pixels = image.getdata()  # Get pixel data
        # Adjust pixel value scaling for better ASCII coverage
        # Ensuring index remains within bounds:
        ascii_str = ''.join(ASCII_CHARS[min(pixel // (256 // len(ASCII_CHARS)), len(ASCII_CHARS) - 1)] for pixel in pixels)
        # Create ASCII image by joining characters and splitting into lines
        ascii_img = '\n'.join(ascii_str[i:i+new_width] for i in range(0, len(ascii_str), new_width))
        print(ascii_img)  # Print the ASCII art
        with open('ascii_image.txt', 'w') as f:  # Write the ASCII art to a text file
            f.write(ascii_img)
    except Exception as e:
        print(f"Unable to process image: {e}")  # Handle exceptions

# Specify the path to your image file
image_path = 'aura4.jpeg'  # Ensure this path is correct based on where the image is stored
process_image(image_path)
```
